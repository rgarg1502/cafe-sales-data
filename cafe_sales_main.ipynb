```python
df = df.replace(['UNKNOWN','ERROR'], np.nan)
```

```python
unique_per_column = {col: df[col].unique() for col in df.columns if col != 'Transaction ID'}
unique_per_column
```

```python
pd.reset_option('display.max_rows')
pd.reset_option('display.max_columns')
pd.reset_option('display.max_colwidth')
```

```python
df.head()
```

```python
float_columns = ['Quantity','Price Per Unit', 'Total Spent']
df[float_columns] = df[float_columns].astype(float)
```

```python
df['Total Spent'].isna().sum()
```

```python
df['Total Spent'] = df['Total Spent'].fillna((df['Quantity'] * df['Price Per Unit']))
```

```python
df['Total Spent'].isna().sum()
```

```python
df.head()
```

```python
df['Price Per Unit'].isna().sum()
```

```python
df['Price Per Unit'] = df['Price Per Unit'].fillna(df['Total Spent'] / df['Quantity'])
```

```python
df['Price Per Unit'].isna().sum()
```

```python
df['Quantity'].isna().sum()
```

```python
df['Quantity'] = df['Quantity'].fillna(df['Total Spent'] / df['Price Per Unit'])
```

```python
df['Quantity'].isna().sum()
```

```python
df.dropna(subset=['Quantity','Total Spent', 'Price Per Unit'], inplace=True)
```

```python
df
```

```python
df['Transaction Date'] = pd.to_datetime(df['Transaction Date'])
```

```python
df['Transaction Date'].isna().sum()
df.shape
```

```python
sorted_df = df.sort_values('Transaction ID')
sorted_df
```

```python
sorted_df['Transaction Date'] = sorted_df['Transaction Date'].ffill(limit=1)
sorted_df['Transaction Date'].isna().sum()
```

```python
sorted_df['Transaction Date'].unique()
```

```python
sorted_df['Transaction Date'] = sorted_df['Transaction Date'].bfill(limit=1)
sorted_df['Item'].isna().sum()
```

```python
price_map = {2.00:'Coffee', 1.50:'Tea', 1.00: 'Cookie', 5.00: 'Salad'}
missing_items = sorted_df['Price Per Unit'].map(price_map)
sorted_df['Item'] = sorted_df['Item'].fillna(missing_items)
sorted_df['Item'].isna().sum()
sorted_df['Item'].unique()
```

```python
mapping = ['Sandwich','Smoothie']
price_4_filter = (sorted_df['Item'].isna()) & (sorted_df['Price Per Unit'] == 4.00)
missing_count_4 = price_4_filter.sum()
if missing_count_4 > 0:
    sorted_df.loc[price_4_filter, 'Item'] = np.random.choice(mapping, size = missing_count_4, p = [0.5, 0.5])
sorted_df['Item'].isna().sum()
```

```python
mapping_3 = ['Cake','Juice']
price_3_filter = (sorted_df['Item'].isna()) & (sorted_df['Price Per Unit'] == 3.00)
missing_count_3 = price_3_filter.sum()
if missing_count_3 > 0:
    sorted_df.loc[price_3_filter, 'Item'] = np.random.choice(mapping_3, size = missing_count_3, p = [0.5, 0.5])
sorted_df['Item'].isna().sum()
```

```python
sorted_df.head(3)
```

```python
sorted_df['Payment Method'].isna().sum()
missing_payment_filter = sorted_df['Payment Method'].isna()
sorted_df.loc[missing_payment_filter,'Payment Method'] = sorted_df['Payment Method'].dropna().sample(sorted_df['Payment Method'].isna().sum()).values
sorted_df['Payment Method'].isna().sum()
```

```python
sorted_df['Location'].isna().sum()
missing_location_filter = sorted_df['Location'].isna()
sorted_df.loc[missing_location_filter,'Location'] = sorted_df['Location'].dropna().sample(sorted_df['Location'].isna().sum()).values
```

```python
sorted_df.isna().sum()
```

```python
sorted_df.to_csv('cleaned_cafe_sales.csv')
```
